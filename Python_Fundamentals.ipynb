Q8.3. Now write a function increment_counter() that tries to do counter = counter + 1 without using the global
keyword. Run it and explain the error you get.

```python
counter = 0
def increment_counter():
    counter = counter + 1
increment_counter()
#UnboundLocalError: local variable 'counter' referenced before assignment
#bina global keyword ke function global variable ko update nahi kar sakta.
```

counter variable.

Q11. Class Attendance Counter
