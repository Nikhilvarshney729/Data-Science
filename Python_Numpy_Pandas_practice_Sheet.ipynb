38. Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard
deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using
np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain
the statistical effects of boundary replacement.

```python
import numpy as np

np.random.seed(42)
arr = np.random.normal(50, 12, 100)

print("Outliers:", np.sum((arr < 30) | (arr > 70)))

cleaned = np.where(arr < 30, 30, np.where(arr > 70, 70, arr))

for name, data in [("Original", arr), ("Cleaned", cleaned)]:
    print(name)
    print("Mean:", np.mean(data))
    print("Variance:", np.var(data))
    print("Std:", np.std(data))
    print("Min:", np.min(data))
    print("Max:", np.max(data))
```

40. Series data integrity: Preserve the required index order, identify missing student records, and filter valid
scores of at least 70. Convert the filtered Series to a dictionary, reconstruct it as a Series, and verify that labelled
values match. Explain why missing records should not automatically be zero marks.
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)

```python
import pandas as pd

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores, index=required)
print(s)

print("Missing:", s[s.isna()])

filtered = s[s >= 70]
d = filtered.to_dict()
new_s = pd.Series(d)

print("Dictionary:", d)
print("New Series:\n", new_s)
print("Match:", filtered.equals(new_s))
```
