26. Create a five-row DataFrame named course_df with columns Student, Course, and Score. Then, in
sequence: display its column names, retrieve Student, create a DataFrame containing Student and Score,
and add a new column named Batch with the same value for all rows.

```python
course_df = pd.DataFrame({
    "Student":["A","B","C","D","E"],
    "Course":["Math","Sci","Hist","Eng","CS"],
    "Score":[85,92,78,88,95]
})

print("Columns:", course_df.columns)          # column names
print("Students:\n", course_df["Student"])    # retrieve Student
print("\nStudent & Score:\n", course_df[["Student","Score"]])  # subset
course_df["Batch"] = "2026"                   # add new column
print("\nFinal:\n", course_df)
```

27. Create a five-row DataFrame named travel_df. Add a column named Code using np.arange(5). Next,
replace that column with a Series containing values only at indexes 1 and 4. Display the DataFrame after each
assignment.

```python
import pandas as pd
import numpy as np
travel_df = pd.DataFrame({
    "Place":["Delhi","Agra","Goa","Jaipur","Mumbai"],
    "Days":[3,2,5,4,6]
})
print("Created:\n", travel_df)
travel_df["Code"] = np.arange(5)
print("\nAfter np.arange:\n", travel_df)
travel_df["Code"] = pd.Series({1:101, 4:202})
print("\nAfter replacement:\n", travel_df)
```
