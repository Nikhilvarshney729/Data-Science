12.Create an array and demonstrate the difference between a slice/view and a copy.

```python
import numpy as np
arr = np.arange(1,6)
# Slice (view)
view_arr = arr[1:4]
view_arr[0] = 999
# Copy
copy_arr = arr[1:4].copy()
copy_arr[0] = 555
print("Slice array:", arr)
print("Original after modifying copy:", arr)
print("Copy array:", copy_arr)
```

16.Create a 5 × 5 matrix containing 1–25 and extract the first row, last column, middle element, and top-left 3 × 3 matrix.

```python
import numpy as np
mat = np.arange(1, 26).reshape(5, 5)
first_row = mat[0]
last_col = mat[:, -1]
middle_elem = mat[2, 2]
top_left_3x3 = mat[0:3, 0:3]
print("First row:", first_row)
print("Last column:", last_col)
print("Middle element:", middle_elem)
print("Top-left 3x3:\n", top_left_3x3)
```
