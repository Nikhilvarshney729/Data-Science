Q29. Use repeat_tests.csv to create a pivot_table() containing both mean and maximum Marks for every Student_ID and Subject combination. Reshape the result with stack() or unstack() to expose the hierarchical index structure, and restore the original pivot-table layout.

```python
repeat_tests = pd.read_csv("repeat_tests.csv")
result = repeat_tests.pivot_table(index="Student_ID", columns="Subject", values="Marks", aggfunc=['mean', 'max'])
result = result.stack()
result = result.unstack()
result
```

Q30. Merge students_master.csv with marks_long.csv on Student_ID. Use pivot() to produce a wide table indexed by Student_ID and Student_Name with Python, DBMS and Maths columns; then use reset_index() to return the identifiers to ordinary columns. Explain why this pivot() succeeds while pivot() on repeat_tests.csv does not.

```python
student_master = pd.read_csv("students_master.csv")
marks_long = pd.read_csv("marks_long.csv")
result = pd.merge(student_master, marks_long, on='Student_ID', how='inner')
result = result.pivot(index=['Student_ID', 'Student_Name'], columns='Subject', values='Marks')
result = result.reset_index()
result
```
