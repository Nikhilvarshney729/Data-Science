# **Section B**

16. For each branch, calculate the total, average, minimum, maximum and standard deviation of units sold using a single agg() operation.

```python
result16 = df.groupby('branch')['units_sold'].agg(['sum', 'mean', 'min', 'max', 'std'])
print(result16)
```

17. For each product, calculate the total units sold, average discount percentage, maximum customer rating and minimum delivery time using a single aggregation operation.

```python
result17 = df.groupby('product').agg({'units_sold': 'sum', 'discount_pct': 'mean', 'customer_rating': 'max', 'delivery_minutes': 'min'})
print(result17)
```

18. For each branch, calculate the total number of sales records using size() and the number of non-missing customer ratings using count(). Compare the two results.

```python
result18a = df.groupby('branch').size()
result18b = df.groupby('branch')['customer_rating'].count()
print(result18a.compare(result18b))
```

19. Define a user-defined aggregation function to calculate the range (maximum - minimum) of units sold for each product.

```python
result19 = df.groupby('product')['units_sold'].agg(range_units_sold=lambda x: x.max() - x.min())
print(result19)
```

20. Group the dataset by branch and product, calculate the total units sold and total target units, and return the result as a DataFrame using as_index=False.

```python
result = df.groupby(['branch', 'product'], as_index=False).agg({
    'units_sold': 'sum',
    'target_units': 'sum'
})

print(result)
```

21. Iterate over the groups created using the branch column and display the branch name, its corresponding sales records and total units sold.

```python
for branch, group in df.groupby('branch'):
    print("Branch:", branch)
    print(group[['product', 'units_sold']])
    print("Total Units Sold:", group['units_sold'].sum())
    print()
```

**Section C**

26. Identify missing values in customer_rating and replace them with the average customer rating of the corresponding product within the same branch using groupby() and transform(). Display the dataset before and after replacement.

```python
result26 = df.groupby(['branch', 'product'])['customer_rating'].transform(lambda x: x.fillna(x.mean()))
print(result26)
```

27. Calculate the weighted average discount percentage for each branch, considering units_sold as the weight. Use a user-defined function with groupby() and apply(). Calculate the weighted average as the sum of (units sold × discount percentage) divided by the total units sold.

```python
result27 = df.groupby('branch').apply(
    lambda x: (x['units_sold'] * x['discount_pct']).sum()
    / x['units_sold'].sum(),
    include_groups=False
).reset_index(name='weighted_avg_discount_pct')

print(result27)
```

28. Identify the top three sales records with the highest units sold for every branch using groupby() and apply(). Display the branch, product, date and units sold for the selected records.

```python
result28 = df.groupby('branch').apply(
    lambda x: x.nlargest(3, 'units_sold'),
    include_groups=False
).reset_index(level=0)

print(result28[['branch', 'product', 'day', 'units_sold']])
```

29. Calculate the percentage contribution of each sales record to the total units sold by its respective branch using transform(). Create a new column named sales_contribution_pct and verify that the total contribution for each branch is 100% (subject to rounding).

```python
df['sales_contribution_pct'] = (
    df['units_sold'] /
    df.groupby('branch')['units_sold'].transform('sum')
) * 100

print(df[['branch', 'product', 'units_sold', 'sales_contribution_pct']])

print(df.groupby('branch')['sales_contribution_pct'].sum())
```

30. Divide all sales records into four quartile-based categories using pd.qcut() on units_sold: Low Sales, Below Average Sales, Above Average Sales and High Sales. Group by these categories and calculate total units sold, average discount percentage, average customer rating, average delivery time and number of sales records.

```python
result30 = pd.qcut(df['units_sold'], q=4, labels=['Low Sales', 'Below Average Sales', 'Above Average Sales', 'High Sales'])
print(result30)
```
